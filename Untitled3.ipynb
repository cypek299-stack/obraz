```python



import torch
import torch.nn as nn
import torchvision
import torchvision.transforms as transforms
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

from torch.utils.data import DataLoader
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
from collections import Counter



torch.manual_seed(42)
np.random.seed(42)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Urządzenie:", device)

BATCH_SIZE = 128
EPOCHS = 8
LEARNING_RATE = 0.001



transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.5,), (0.5,))
])

train_dataset = torchvision.datasets.FashionMNIST(
    root="./data",
    train=True,
    download=True,
    transform=transform
)

test_dataset = torchvision.datasets.FashionMNIST(
    root="./data",
    train=False,
    download=True,
    transform=transform
)


generator = torch.Generator().manual_seed(42)

train_dl = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    generator=generator,
    num_workers=2,
    pin_memory=torch.cuda.is_available()
)

test_dl = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    pin_memory=torch.cuda.is_available()
)

classes = [
    "T-shirt/top", "Trouser", "Pullover", "Dress", "Coat",
    "Sandal", "Shirt", "Sneaker", "Bag", "Ankle boot"
]

print("Liczba obrazów treningowych:", len(train_dataset))
print("Liczba obrazów testowych:", len(test_dataset))
print("Liczba klas:", len(classes))


images, labels = next(iter(train_dl))

fig, axes = plt.subplots(2, 5, figsize=(12, 5))

for i, ax in enumerate(axes.flat):
    img = images[i].squeeze() * 0.5 + 0.5
    ax.imshow(img, cmap="gray")
    ax.set_title(classes[labels[i].item()])
    ax.axis("off")

plt.suptitle("Przykładowe obrazy FashionMNIST")
plt.tight_layout()
plt.show()



class MLP(nn.Module):
    def __init__(self, hidden=(256, 128)):
        super().__init__()

        layers = [nn.Flatten()]
        in_dim = 28 * 28

        for h in hidden:
            layers.append(nn.Linear(in_dim, h))
            layers.append(nn.ReLU())
            in_dim = h

        layers.append(nn.Linear(in_dim, 10))
        self.net = nn.Sequential(*layers)

    def forward(self, x):
        return self.net(x)


def count_params(model):
    return sum(p.numel() for p in model.parameters())




def evaluate(model, dataloader):
    model.eval()
    total_loss = 0.0
    correct = 0
    total = 0
    criterion = nn.CrossEntropyLoss(reduction="sum")

    with torch.no_grad():
        for xb, yb in dataloader:
            xb = xb.to(device)
            yb = yb.to(device)

            outputs = model(xb)
            total_loss += criterion(outputs, yb).item()

            predictions = outputs.argmax(dim=1)
            correct += (predictions == yb).sum().item()
            total += yb.size(0)

    return total_loss / total, correct / total



architectures = {
    "MLP (128,)": (128,),
    "MLP (256, 128)": (256, 128),
    "MLP (512, 256, 128)": (512, 256, 128)
}

results = []
histories = {}
models = {}

for name, hidden in architectures.items():

    print("\n" + "=" * 60)
    print("Trening modelu:", name)


    torch.manual_seed(42)

    model = MLP(hidden=hidden).to(device)
    optimizer = torch.optim.Adam(
        model.parameters(),
        lr=LEARNING_RATE
    )
    criterion = nn.CrossEntropyLoss()

    history = {
        "train_loss": [],
        "test_loss": [],
        "test_acc": []
    }

    for epoch in range(EPOCHS):

        model.train()
        running_loss = 0.0

        for xb, yb in train_dl:
            xb = xb.to(device)
            yb = yb.to(device)

            optimizer.zero_grad()

            outputs = model(xb)
            loss = criterion(outputs, yb)

            loss.backward()
            optimizer.step()

            running_loss += loss.item() * xb.size(0)

        train_loss = running_loss / len(train_dataset)
        test_loss, test_acc = evaluate(model, test_dl)

        history["train_loss"].append(train_loss)
        history["test_loss"].append(test_loss)
        history["test_acc"].append(test_acc)

        print(
            f"Epoka {epoch + 1:02d}/{EPOCHS} | "
            f"Train loss: {train_loss:.4f} | "
            f"Test loss: {test_loss:.4f} | "
            f"Test accuracy: {test_acc * 100:.2f}%"
        )

    params = count_params(model)
    final_loss, final_acc = evaluate(model, test_dl)

    results.append({
        "Architektura": name,
        "Liczba parametrów": params,
        "Accuracy [%]": final_acc * 100
    })

    histories[name] = history
    models[name] = model



results_df = pd.DataFrame(results)
results_df = results_df.sort_values(
    "Accuracy [%]", ascending=False
).reset_index(drop=True)

print("\nPORÓWNANIE ARCHITEKTUR")
display(
    results_df.style.format({
        "Liczba parametrów": "{:,}",
        "Accuracy [%]": "{:.2f}%"
    })
)

best_name = results_df.iloc[0]["Architektura"]
best_model = models[best_name]
best_history = histories[best_name]

print("\nNajlepszy model:", best_name)
print(
    f"Accuracy najlepszego modelu: "
    f"{results_df.iloc[0]['Accuracy [%]']:.2f}%"
)
print(
    f"Liczba parametrów: "
    f"{count_params(best_model):,}"
)



epochs = range(1, EPOCHS + 1)

fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))

axes[0].plot(
    epochs, best_history["train_loss"],
    marker="o", label="Strata treningowa"
)
axes[0].plot(
    epochs, best_history["test_loss"],
    marker="o", label="Strata testowa"
)
axes[0].set_xlabel("Epoka")
axes[0].set_ylabel("Cross-entropy loss")
axes[0].set_title(f"Krzywe uczenia: {best_name}")
axes[0].legend()
axes[0].grid(alpha=0.3)

axes[1].plot(
    epochs,
    np.array(best_history["test_acc"]) * 100,
    marker="o"
)
axes[1].set_xlabel("Epoka")
axes[1].set_ylabel("Accuracy [%]")
axes[1].set_title("Accuracy na zbiorze testowym")
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()



best_model.eval()

all_images = []
all_labels = []
all_preds = []

with torch.no_grad():
    for xb, yb in test_dl:
        outputs = best_model(xb.to(device))
        preds = outputs.argmax(dim=1).cpu()

        all_images.append(xb.cpu())
        all_labels.append(yb.cpu())
        all_preds.append(preds)

all_images = torch.cat(all_images)
all_labels = torch.cat(all_labels)
all_preds = torch.cat(all_preds)

correct_idx = torch.where(all_preds == all_labels)[0].tolist()
wrong_idx = torch.where(all_preds != all_labels)[0].tolist()

print("\nPoprawne predykcje:", len(correct_idx))
print("Błędne predykcje:", len(wrong_idx))


selected = correct_idx[:6] + wrong_idx[:6]

fig, axes = plt.subplots(2, 6, figsize=(15, 6))

for ax, idx in zip(axes.flat, selected):

    img = all_images[idx].squeeze() * 0.5 + 0.5

    true_label = all_labels[idx].item()
    pred_label = all_preds[idx].item()

    correct = true_label == pred_label

    ax.imshow(img, cmap="gray")
    ax.set_title(
        f"Pred.: {classes[pred_label]}\n"
        f"Praw.: {classes[true_label]}",
        color="green" if correct else "red",
        fontsize=8
    )
    ax.axis("off")

plt.suptitle("Przykładowe predykcje najlepszego modelu")
plt.tight_layout()
plt.show()


cm = confusion_matrix(
    all_labels.numpy(),
    all_preds.numpy(),
    labels=list(range(10))
)

fig, ax = plt.subplots(figsize=(10, 8))

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=classes
)

disp.plot(
    ax=ax,
    cmap="Blues",
    xticks_rotation=45,
    values_format="d",
    colorbar=False
)

ax.set_title(f"Macierz pomyłek: {best_name}")
plt.tight_layout()
plt.show()



errors = Counter()

for true_label, pred_label in zip(
    all_labels.tolist(),
    all_preds.tolist()
):
    if true_label != pred_label:
        errors[(true_label, pred_label)] += 1

print("\n10 najczęstszych pomyłek:")

for (true_label, pred_label), count in errors.most_common(10):
    print(
        f"{classes[true_label]} -> {classes[pred_label]}: "
        f"{count} przypadków"
    )

if errors:
    (true_label, pred_label), count = errors.most_common(1)[0]

    print("\nNajczęstsza pomyłka kierunkowa:")
    print(
        f"{classes[true_label]} -> {classes[pred_label]} "
        f"({count} przypadków)"
    )



print("\n" + "=" * 60)
print("PODSUMOWANIE EKSPERYMENTU")
print("=" * 60)

print("Zbiór danych: FashionMNIST")
print("Liczba epok:", EPOCHS)
print("Współczynnik uczenia:", LEARNING_RATE)
print("Najlepsza architektura:", best_name)
print(
    f"Accuracy testowe: "
    f"{results_df.iloc[0]['Accuracy [%]']:.2f}%"
)
print(
    f"Parametry najlepszego modelu: "
    f"{count_params(best_model):,}"
)

last_train_loss = best_history["train_loss"][-1]
last_test_loss = best_history["test_loss"][-1]

print(f"Końcowa strata treningowa: {last_train_loss:.4f}")
print(f"Końcowa strata testowa: {last_test_loss:.4f}")

if last_test_loss > last_train_loss:
    print(
        "Strata testowa jest większa od treningowej. "
        "Może to wskazywać na lukę generalizacji."
    )
else:
    print(
        "Strata testowa nie jest większa od treningowej. "
        "Należy ocenić również przebieg krzywych w kolejnych epokach."
    )

print("\nTabela wyników do sprawozdania:")
print(results_df.to_string(index=False))
```
